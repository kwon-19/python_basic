## 3. 딕셔너리 쌍 추가, 삭제하기

```python
a = {1 : 'a'}

a[2] = 'b'
a
```

```python
a['name'] ='pey'
a
```

```python
a[3] = [1, 2, 3]
a
```

```python
# del 키워드로 요소 삭제하기
del a[1]
a
```

```python
del a['name']
a
```

### 딕셔너리 만들 때 주의 사항

```python
a = {1: 'a', 1: 'b'} # key는 고유한 값(중복 불가)
a
```

```python
a = {[1,2]: 'hi'} # key에 리스트 사용 불가
a
```

```python
a = {(1,2):'hi'}
a
```

```python
a[(1,2)]
```

## 5 딕셔너리 관련 함수

### 리스트 만들기 : keys()

```python
a = {'name' : 'kig', 'phone' : '010-1234-1234', 'birth' : 1118}
```

```python
a.keys()
```

```python
type(a.keys())
```

```python
list(a.keys())
```

```python
type(list(a.keys()))
```

```python
for k in a.keys():
    print(k)
```

### value 리스트 만들기 : values()

```python
a.values()
```

### key, value 쌍 얻기 : items()

```python
a.items()
```

### key : value 쌍 모두 지우기 : clear()

```python
a.clear()
a
```

### key로 value얻기 : get()

```python
a = {'name' : 'kig', 'phone' : '010-1234-1234', 'birth' : '1118'}

a.get('phone')
```

```python
print(a.get('age'))
```

```python
print(a[age])
```
