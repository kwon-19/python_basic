### break문

```python
# 커피 자판기
coffee = 10
money = 300

while money:
    print('돈을 받았으니 커피를 줍니다.')
    coffee -= 1
    print('남은 커피는 %d개 입니다.'%coffee)
    if coffee == 0:
        print('커피가 다 떨어졌습니다. 판매를 중지합니다.')
        break # 반복문 강제 종료
  
```

```python
coffee = 10

while True:
    money = int(input('돈을 넣어주세요.'))
    if money == 300:
        print('커피를 줍니다.')
        coffee -= 1
    elif money > 300:
        print("거스름돈 %d를 주고, 커피를 줍니다."%(money - 300))
        coffee -= 1
    else:
        print('돈을 다시 돌려주고, 커피를 주지 않습니다.')
        print('남은 커피의 양은 %d개 입니다.'%coffee)

    if coffee == 0:
        print('커피가 다 떨어졌습니다. 판매를 중지합니다.')
        break
```

총 5명의 학생이 시험을 보았는데 시험 점수가 60점 이상이면 합격이고 그렇지 않으면 불합격이다. 합격인지, 불합격인지 결과를 보여 주시오.

```python
marks = [90, 25, 67, 45, 80]
num = 0

for mark in marks:
    num += 1
    if mark >= 60:
        print('%d번 학생은 합격입니다.'%num)
    else:
        print('%d번 학생은 불합격입니다.'%num)
```

```python
marks = [90, 25, 67, 45, 80]
num = 0

for mark in marks:
    num += 1
    if mark <60: continue
    print(f"{num}번 학생 축하합니다. 합격입니다.")
```

### 리스트 컴프리헨션(list comprehension)

```python
# 일반적은 for문
a = [1,2,3,4]
result = []

for num in a:
    result.append(num*3)
print(result)
```

```python
result = [num*4 for num in a]
print(result)
```

```python
# 리스트 컴프리헨션 안에 if문 추가
result = [num*2 for num in a if num%2 == 0]
print(result)
```

```python
#리스트컴프리헨션으로 구구단
result = [x*y for x in range(2,10)
                for y in range(1,10)]

print(result)
```
