## 2. 리스트를 복사하고자 할 때

```python
a = [1, 2, 3]
b = a
b
```

```python
id(a)
```

```python
id(b)
```

```python
a[1] = 4
a
```

```python
b
```

```python
a is b  # is : 동일한 데이터(객체)를 가르키고 있는지 판단
```

### b변수를 생성할 때 a 변수의 값을 가져오면서 a와는 다른 주소를 가리키도록 만드는 방법

```python
# 방법1) [:]
a = [1, 2, 3]
b = a[:]
print(a,b)
```

```python
a is b
```

```python
id(a)
```

```python
id(b)
```

```python
a[1] = 4
a
```

```python
b
```

```python
b = a[:]
b
```

```python
# 방법2) copy 모듈 이용하기
from copy import copy

a = [1, 2, 3]
b = copy(a)
print(a,b)
```

```python
a is b
```

```python
print(id(a), id(b))
```

```python
# 방법3) copy() 함수 사용하기
a = [1, 2, 3]
b = a.copy()

a is b
```

## 3. 변수를 만드는 여러가지 방법

### 튜플로 변수 만들기

```python
a, b = ('python', 'life')
```

```python
a
```

```python
b
```

```python
(a, b) = 'python', 'life'
```

```python
a
```

```python
b
```

### 리스트로 변수 만들기

```python
[a, b] = ['python', 'life']
```

```python
a
```

```python
b
```

여러 개의 변수에 같은 값을 대입

```python
a = b = c = 'python'

print(a, b, c)
```

두 변수의 값 바꾸기

```python
# 일반적인 프로그래밍에서 두 변수 값 교체할 때
a = 3
b = 5
print(a, b)
```

```python
temp = a
a = b
b = temp

print(a, b)
```

```python
# python에서
a = 3
b = 5
a, b = b, a

print(a, b)
```
