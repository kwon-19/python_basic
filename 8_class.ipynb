## 1. 클래스는 왜 필요한가?

```python
# 계산기 프로그램을 함수로 만들기
result = 0

def add(num):
    global result
    result += num
    return result

add(3)
```

```python
add(4)
```

```python
# 계산기 2대가 필요한 상황

result1 = 0
result2 = 0

def add1(num):
    global result1
    result1 += num
    return result1

def add2(num):
    global result2
    result2 += num
    return result2

add1(3)
print(add1(4))

add2(3)
print(add2(7))
```

```python
# 클래스로 해결

class Calculator:
    def __init__(self):  # 생성자 함수(클래스 변수 초기화)
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result
        
```

4. 더하기 기능 만들기

```python
class FourCal:
    def setdata(self, first, second): # 숫자 2개 입력받는 함수(메서드)
        self.first = first
        self.second = second

    def add(self):  # 더하기 함수
        result = self.first + self.second
        return result
```

5) 곱하기, 빼기, 나누기 기능 만들기

```python
class FourCal:
    def setdata(self, first, second): # 숫자 2개 입력받는 함수(메서드)
        self.first = first
        self.second = second

    def add(self):  # 더하기 함수
        result = self.first + self.second
        return result

    def mul(self):  # 곱하기 함수
        result = self.first * self.second
        return result
    
    def sub(self):  # 빼기 함수
        result = self.first - self.second
        return result

    def div(self):  # 나누기 함수
        result = self.first / self.second
        return result
```

```python
a = FourCal()
b = FourCal()

a.setdata(4,2)
b.setdata(3,8)
```

```python
a.add()
```

```python
a.mul()
```

```python
a.div()
```

```python
b.add()
```

```python
b.mul()
```

```python
b.sub()
```

```python
b.div()
```

## 4. 생성자(constructor)

```python
a = FourCal()
a.add()
```

```python
class FourCal:
    def __init__(self, first, second):  # 생성자 함수
        self.first = first
        self.second = second
    
    def add(self):  # 더하기 함수
        result = self.first + self.second
        return result

    def mul(self):  # 곱하기 함수
        result = self.first * self.second
        return result
    
    def sub(self):  # 빼기 함수
        result = self.first - self.second
        return result

    def div(self):  # 나누기 함수
        result = self.first / self.second
        return result
```

```python
a = FourCal()
```

```python
# FourCal 클래스로 객체가 생성되면 생성자 함수에게 매개변수값이 전달된다.
a = FourCal(4,2)
```

```python
a.first
```

```python
a.second
```

```python
a.add()
```

```python
a.div()
```

## 5. 클래스의 상속

```python
# FourCal 클래스를 상속받아 MoreFourCal 클래스 선언
class MoreFourCal(FourCal):
    pass
```

```python
a = MoreFourCal(4,2)
```

```python
a.add()
```

```python
a.sub()
```

```python
class MoreFourCal(FourCal):
    def pow(self):  # 기능 추가
        result = self.first ** self.second
        return result
```

```python
a = MoreFourCal(4,2)
a.pow()
```

```python
a.mul()
```

## 6. 메서드 오버라이딩(재정의)

```python
a = FourCal(4,0)
a.add()
```

```python
a.div()
```

```python
class SafeFourCal(FourCal):
    def div(self):  # FourCal클래스 div메서드 재정의
        if self.second ==0:
            return 0
        else:
            result = self.first / self.second
            return result
```

```python
a = SafeFourCal(4,0)
a.div()
```
