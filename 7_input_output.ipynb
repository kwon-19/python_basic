## 파일 읽고 쓰기
1. 파일 생성하기

```python
f = open('newfile.txt','w')
f.close()
```

```python
# 특정 경로(절대경로)에 파일 생성
f = open('D:/kig/python/newfile.txt','w')
f.close()
```

2. 파일을 쓰기 모드로 열어 내용 작성하기

```python
f = open('newfile.txt','w')

for i in range(1,11):
    data = f'{i}번째 줄입니다.\n'
    f.write(data)

f.close()
```

3. 파일을 읽는 여러가지 방법

```python
# 한 줄 읽기가 readline()
f = open('newfile.txt','r')

line = f.readline()
print(line)

f.close()
```

```python
f = open('newfile.txt','r')

while True:
    line = f.readline()
    if not line: break
    print(line)

f.close()
```

```python
# 여러줄 읽기 : readlines()
f = open('newfile.txt','r')

lines = f.readlines()
print(line)

for line in lines:
    print(line)

f.close()
```

```python
# 줄바꿈(\n)문자 제거하기
f = open('newfile.txt','r')

lines = f.readlines()

for line in lines:
    print(line.strip())

f.close()
```

```python
# 파일 전체 내용 읽기 : read()
f = open('newfile.txt','r')

data = f.read()
print(data)

f.close()
```

```python
# 파일 객체를 for문과 함께 사용하기
f = open('newfile.txt','r')

for line in f:
    print(line.strip())

f.close()
```

4) 파일에 새로운 내용 추가하기

```python
f = open('newfile.txt','a')

for i in range(11,21):
    data = f'{i}번째 줄입니다.\n'
    f.write(data)

f.close()
```

5) with문과 함께 사용하기

```python
with open('foo.txt','w') as f:
    f.write('Life is too short, You need python')
```
